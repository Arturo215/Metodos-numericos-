# Método de bisección

Queremos resolver la ecuación $x^3+4x-10=0$.

Para resolver esta ecuación se va a ocupar el **método de bisección**. Para esto requerimos confirmar que la función $f(x)$ evaluada en los extremos opuestos del intervalo $[a,b]$ tiene signos opuestos es decir $f(a)*f(b)<0$.

En la siguiente gráfica podemos apreciar a la función $F(x)= x^3+x^2-10$ y a los puntos A y B. Destacando que la función corta al eje x por el cero, es decir tiene una raíz y que los puntos A,B al ser evaluados en la función tienen signo opuesto.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definimos la función basada en tu ecuación x^3 + 4x - 10 = 0
def f(x):
    return x**3 + 4 * x - 10

# Definimos los puntos del intervalo [a, b]
a = 1
b = 2

# Creamos un rango de valores para x para graficar la curva de manera continua
x_vals = np.linspace(0, 3, 400)
y_vals = f(x_vals)

# Configuramos la figura y los ejes
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label='$f(x) = x^3 + 4x - 10$', color='blue', linewidth=2)
plt.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.7) # Línea del eje X

# Graficamos los puntos A y B
plt.plot(a, f(a), 'ro', markersize=8, label=f'Punto A ({a}, {f(a)})')
plt.plot(b, f(b), 'go', markersize=8, label=f'Punto B ({b}, {f(b)})')

# Añadimos etiquetas, título y leyenda
plt.title('Gráfica para el Método de Bisección', fontsize=12)
plt.xlabel('x', fontsize=10)
plt.ylabel('f(x)', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

# Mostramos la gráfica
plt.show()
```

**Desarrollo del método**
Consiste en partir el intervalo (donde sabemos que está la raíz gracias al TVI) a la mitad para encontrar a p, y al hacer $f(p)$ podremos saber en que mitad está la raíz ya sea entre $f(p)$ y $f(b)$, o $f(a)$ y $f(p)$. De tal manera que al iterar este proceso, eventualmente llegaremos a la solución buscada.
La solución para $f(x)$ se encuentra en el siguiente código.

```python
# definimos la función que vamos a trabajar
def f(x: int) -> int:
    return x**3 + (4 * x**2) - 10
# definimos las contantes necesarias para calcular los valores
def main() -> None:
    a = 1
    b = 2
    TOL = 1e-4 #constante de tolerancia
    N = 20

    FA = f(a)
    FB = f(b)
# condicion para saber si hay una raíz en el intervalo dado
    if FA * FB >= 0:
        print("Error, f(a) y f(b) tienen signos iguales")
        return
# bucle de iteración para hallar las raíz
    i = 0
    while i <= N:
        aux = (b - a) / 2 # Desde aqui
        p = a + aux
        FP = f(p)

        if FP == 0 or aux < TOL:
            print(f"La raiz apoximada en p es = {p}, hallada en la iteracion {i}")
            break

        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p
        i += 1 # Hasta aca

if __name__ == "__main__":
    main()
```
